```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('../gc_stats.csv')
df
```

```python
sg_rolling = np.array([sum(df['synapsegene'][:i]) for i in range(1, len(df['synapsegene']) + 1)])
ng_rolling = np.array([sum(df['neurongene'][:i]) for i in range(1, len(df['synapsegene']) + 1)])
so_rolling = np.array([sum(df['synapse_obj'][:i]) for i in range(1, len(df['synapsegene']) + 1)])
no_rolling = np.array([sum(df['neuron_obj'][:i]) for i in range(1, len(df['synapsegene']) + 1)])
index = [i for i in range(5)]
```

```python
plt.plot(index, df['gc_gen2'])
plt.plot(index, sg_rolling, label = 'sg count')
plt.plot(index, so_rolling, label = 'so count')
plt.plot(index, ng_rolling, label = 'ng count')
plt.plot(index, no_rolling, label = 'no count')

plt.plot(index, sg_rolling + ng_rolling + so_rolling + no_rolling, label = 'combined')
plt.title('Gen 2 Objects vs Various Other Object (Counts)')
plt.legend()
```
